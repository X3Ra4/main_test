# Температура за тиждень


```python
from IPython.display import Markdown, display


def show(headers, rows):
    lines = ["| " + " | ".join(headers) + " |",
             "| " + " | ".join(["---"] * len(headers)) + " |"]
    lines.extend("| " + " | ".join(map(str, row)) + " |" for row in rows)
    display(Markdown("\n".join(lines)))

temperatures = [2, 5, 4, 6, 7, 3, 1]
show(["Показник", "Значення, °C"], [
    ("Середня температура", f"{sum(temperatures) / len(temperatures):.2f}"),
    ("Максимальна", max(temperatures)),
    ("Мінімальна", min(temperatures)),
    ("Розмах", max(temperatures) - min(temperatures)),
])
```
