# Результати голосування


```python
from IPython.display import Markdown, display


def show(headers, rows):
    lines = ["| " + " | ".join(headers) + " |",
             "| " + " | ".join(["---"] * len(headers)) + " |"]
    lines.extend("| " + " | ".join(map(str, row)) + " |" for row in rows)
    display(Markdown("\n".join(lines)))

from collections import Counter

votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
counts = Counter(votes)
ranking = sorted(counts.items(), key=lambda pair: (-pair[1], pair[0]))
show(["Фрукт", "Кількість голосів"], sorted(counts.items()))
display(Markdown(f"**Лідер — {ranking[0][0]}**"))
display(Markdown("### Топ-3"))
show(["Місце", "Фрукт", "Голоси"], [(i, fruit, count) for i, (fruit, count) in enumerate(ranking[:3], 1)])
```
