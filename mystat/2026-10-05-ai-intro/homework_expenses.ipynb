# Щоденні витрати


```python
from IPython.display import Markdown, display


def show(headers, rows):
    lines = ["| " + " | ".join(headers) + " |",
             "| " + " | ".join(["---"] * len(headers)) + " |"]
    lines.extend("| " + " | ".join(map(str, row)) + " |" for row in rows)
    display(Markdown("\n".join(lines)))

expenses = [1200, 1500, 800, 600, 1000, 400, 900]
show(["Показник", "Значення"], [
    ("Загальна сума", sum(expenses)),
    ("Середні витрати", f"{sum(expenses) / len(expenses):.2f}"),
    ("Максимальні", max(expenses)),
    ("Мінімальні", min(expenses)),
    ("Дні з витратами понад 1000", sum(amount > 1000 for amount in expenses)),
])
```
