# Оцінки з трьох предметів


```python
from IPython.display import Markdown, display


def show(headers, rows):
    lines = ["| " + " | ".join(headers) + " |",
             "| " + " | ".join(["---"] * len(headers)) + " |"]
    lines.extend("| " + " | ".join(map(str, row)) + " |" for row in rows)
    display(Markdown("\n".join(lines)))

math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]
scores = {"Математика": math, "Фізика": physics, "Англійська": english}
means = {subject: sum(values) / len(values) for subject, values in scores.items()}
show(["Предмет", "Середній бал", "Мінімум", "Максимум"], [
    (subject, f"{means[subject]:.2f}", min(values), max(values))
    for subject, values in scores.items()
])
best = max(means, key=means.get)
print(best)
```
